### 1.1 Preprocessing Steps
To start, I loaded the raw text posts from the `final_posts.json` file, extracted the `post_id` and `text`, and merged them with `emotion.csv` (for labels) and `data_identification.csv` (for train/test split).
After merging, I removed duplicates and cleaned blank rows.

**Cleaning details:**

* Replaced URLs and @mentions with tags (`<URL>`, `<USER>`) so the model could still recognize their presence.

* Normalized repeated letters (e.g., “soooo happy” → “soo happy”).

* For hashtags, I created two versions:

    * v1: Remove “#” but keep the word (#happy → happy).

    * v2: Keep the hashtag mark (#happy → #happy).

* Removed unnecessary whitespace and line breaks.

* Generated an integer label column mapped from emotion names (anger, disgust, fear, joy, sadness, surprise).

This step ensured that all texts were clean, consistent, and still kept emotional signals like emojis and hashtags.
I kept the emojis and hashtags because emojis often explicitly express emotions(😊=joy, 😡=anger...), so removing them would make the text lose its emotional intensity; besides, many users summarize their feelings through hashtags, for eample, “Finally done with exams #happy #relieved”. These hashtags are almost like the extra hints about the emotion category. So instead of removing them, I keep the word (e.g., #happy → happy) so the model still learns the emotional cue, or keep the full hashtag (e.g., remain #happy as it is), allowing the model to differentiate between normal words and hashtags.

In addition, BERT or ELECTRA treat tokens such as “😢” or “#love” as unique embeddings. These embeddings help the model understand the overall sentiment direction, especially when the text is short or sarcastic.


### 1.2 Feature Engineering Steps

Although transformer models already capture context, I added some lightweight emotion-related features to improve robustness.

* Emoji count: Emojis often reveal strong emotions.

* Punctuation counts: Exclamation "!" and question "?" marks show excitement or confusion.

* Uppercase ratio: "ALL CAPS" can express anger or surprise.

* Hashtag and user tag counts: These sometimes indicate topics or interactions.

* URL and mention tags: Show if the post links to something or someone.

After normalizing these features, I trained a simple logistic regression classifier.
Its prediction probabilities were later used in a soft-vote ensemble together with the ELECTRA model output, slightly improving macro-F1 stability.


### 1.3 Explanation of Your Model

I fine-tuned the `electra-small-discriminator` model using the Hugging Face library because it is 3 times faster than BERT but still achieves similar accuracy.(The more detailed explaination in the bonus section.)

**Main training settings:**

* max_length = 96 -> short sequences for faster training.

* batch size = 32 (train) / 64 (eval)

* epochs = 2 -> good balance between speed and performance.

* learning_rate = 3e-5, weight_decay = 0.01.

* Used safetensors=True to avoid old `.bin` security warnings.

**Training pipeline:**

* Tokenized texts with dynamic padding.

* Created a PyTorch dataset using Hugging Face Dataset.

* Trained ELECTRA on the cleaned texts (`text_v1`) and evaluated macro-F1 and accuracy using `evaluate`.

* Generated predictions for the test set and saved to submission_electra.csv.

* Combined ELECTRA probabilities with the feature-engineering Logistic Regression probabilities for a soft-voting ensemble and saved the csv file.

# **Competition Code**

## 1. Preprocessing Steps

```python
import json
import re
import pandas as pd

# Load raw posts from JSON
with open("final_posts.json", "r", encoding="utf-8") as f:
    raw = json.load(f)

def parse_posts(data):
    # Extract post_id and text from JSON into a DataFrame
    rows = []
    for item in data:
        try:
            p = item["root"]["_source"]["post"]
            rows.append({"id": p["post_id"], "text": p["text"]})
        except:
            pass
    return pd.DataFrame(rows)

posts_df = parse_posts(raw)
# Load labels and split info
emo_df  = pd.read_csv("emotion.csv")
id_df   = pd.read_csv("data_identification.csv")
samp_df = pd.read_csv("sampleSubmission.csv") 

# Merge into a single table
df = id_df.merge(posts_df, on="id", how="left").merge(emo_df, on="id", how="left")

# Split into train/test
train_df = df[df["split"].str.lower()=="train"].copy()
test_df  = df[df["split"].str.lower()=="test"].copy()

# Basic normalization and cleaning of whitespaces
train_df["text"] = train_df["text"].fillna("").str.replace(r"\s+"," ", regex=True).str.strip()
test_df["text"]  = test_df["text"].fillna("").str.replace(r"\s+"," ", regex=True).str.strip()

# Lighten the word and keep emotion signals
URL_RE   = re.compile(r'https?://\S+|www\.\S+')
USER_RE  = re.compile(r'@\w+')
HASH_RE  = re.compile(r'#(\w+)') #for example: #happy >> happy
WS_RE    = re.compile(r'\s+')
REP_RE   = re.compile(r'(.)\1{2,}') #for example: soooo >> soo

def clean_text(s, lower=True, url_mode="tag", user_mode="tag", hashtag_mode="keepword",
               num_mode="keep", normalize_repeat=True, strip_punct=False, collapse_ws=True):

    # Keep emotion cues(emoji/#/@/URL) while normalizing noise
    if not isinstance(s, str): return ""
    t = s.lower() if lower else s

    # URLs
    t = URL_RE.sub(" <URL> ", t) if url_mode=="tag" else (URL_RE.sub(" ", t) if url_mode=="drop" else t)
    # @user
    t = USER_RE.sub(" <USER> ", t) if user_mode=="tag" else (USER_RE.sub(" ", t) if user_mode=="drop" else t)
    # #hashtag
    if hashtag_mode=="drop": t = re.sub(r'#\w+',' ',t)
    elif hashtag_mode=="keepword": t = HASH_RE.sub(r'\1', t)
    # numbers
    if num_mode=="tag": t = re.sub(r'\b\d+(\.\d+)?\b', ' <NUM> ', t)
    elif num_mode=="drop": t = re.sub(r'\b\d+(\.\d+)?\b', ' ', t)
    # squash elongations
    if normalize_repeat: t = REP_RE.sub(r'\1\1', t)
    # strip punctuation
    if strip_punct: t = re.sub(r'[^\w\s<>#]', ' ', t)
    # collapse whitespaces
    if collapse_ws: t = WS_RE.sub(' ', t).strip()
    return t

# Two text variants for A/B tests
train_df["text_v1"] = train_df["text"].map(lambda x: clean_text(x, hashtag_mode="keepword"))
test_df["text_v1"]  = test_df["text"].map(lambda x: clean_text(x, hashtag_mode="keepword"))

train_df["text_v2"] = train_df["text"].map(lambda x: clean_text(x, hashtag_mode="keep"))
test_df["text_v2"]  = test_df["text"].map(lambda x: clean_text(x, hashtag_mode="keep"))

#Dedup and drop empty after cleaning
before = len(train_df)
train_df = train_df.drop_duplicates(subset=["text_v1","emotion"]).reset_index(drop=True)
train_df = train_df[train_df["text_v1"].str.len()>0].reset_index(drop=True)
print(f"[Preproc] Dedup: {before} → {len(train_df)}")

# Build label mapping and numeric labels
labels   = sorted(train_df["emotion"].dropna().unique().tolist())
label2id = {l:i for i,l in enumerate(labels)}
id2label = {i:l for l,i in label2id.items()}
train_df["label"] = train_df["emotion"].map(label2id).astype(int)


# Default text column to use(switch to text_v2 to compare)
USE_COL = "text_v1"

print("Classes:", labels)
print("train_df.shape:", train_df.shape, "| test_df.shape:", test_df.shape)
```

## 2. Feature Engineering Steps

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Compute simple statistics from text
EMOJI_RE = re.compile(r'[\U00010000-\U0010ffff]', flags=re.UNICODE)

def fe_row(t: str):
    #Turn a piece of text into a vector of numeric features
    if not isinstance(t, str): t = ""
    L = len(t) or 1
    emojis = EMOJI_RE.findall(t)
    return [
        len(emojis),                          
        t.count('!'),                        
        t.count('?'),                      
        sum(ch.isupper() for ch in t)/L,  
        t.count('#'),                     
        t.count('@'),            
        t.count('<URL>'),           
        t.count('<USER>')       
    ]
fe_cols = ["emoji_cnt","exclam","question","upper_ratio","hash_cnt","user_cnt","url_tag","user_tag"]

# Use cleaned column
train_fe = np.vstack([fe_row(x) for x in train_df[USE_COL]])
test_fe  = np.vstack([fe_row(x) for x in test_df[USE_COL]])

# Standardize
scaler = StandardScaler()
train_fe_sc = scaler.fit_transform(train_fe)
test_fe_sc  = scaler.transform(test_fe)

# Train a light LR classifier
y = train_df["label"].values
clf_fe = LogisticRegression(max_iter=1000, n_jobs=-1, C=2.0)
clf_fe.fit(train_fe_sc, y)

# Get class probabilities for test (for later fusion)
proba_fe = clf_fe.predict_proba(test_fe_sc)
print("[FE] feature probs shape:", proba_fe.shape)
```

## 3. Model Implementation Steps

```python
import torch, evaluate
from datasets import Dataset
from transformers import (
    AutoTokenizer, AutoModelForSequenceClassification,
    TrainingArguments, Trainer, DataCollatorWithPadding
)
from sklearn.metrics import f1_score
import numpy as np

# Model & tokenizer
model_name = "google/electra-small-discriminator"

tok = AutoTokenizer.from_pretrained(model_name, use_fast=True)
def tok_fn(batch):
    # Dynamic padding+shorter max_length for speed
    return tok(batch["text"], truncation=True, max_length=96)

# Build HF datasets
train_ds = Dataset.from_pandas(train_df[[USE_COL, "label"]].rename(columns={USE_COL: "text"}))
test_ds  = Dataset.from_pandas(test_df[[USE_COL]].rename(columns={USE_COL: "text"}))

train_ds = train_ds.map(tok_fn, batched=True, remove_columns=["text"])
test_ds  = test_ds.map(tok_fn, batched=True, remove_columns=["text"])

collator = DataCollatorWithPadding(tokenizer=tok)

# Load classifier with safetensors
model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=len(labels),
    id2label=id2label,
    label2id=label2id,
    use_safetensors=True
)

# Metrics
acc_metric = evaluate.load("accuracy")
f1_metric  = evaluate.load("f1")
def compute_metrics(eval_pred):
    logits, ref = eval_pred
    pred = logits.argmax(-1)
    return {
        "accuracy": acc_metric.compute(predictions=pred, references=ref)["accuracy"],
        "f1_macro": f1_metric.compute(predictions=pred, references=ref, average="macro")["f1"]
    }

# Training args
args = TrainingArguments(
    output_dir="./electra_out",
    learning_rate=3e-5,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=64,
    num_train_epochs=2,
    weight_decay=0.01,
    logging_steps=100,
    report_to="none",
    fp16=torch.cuda.is_available()
)

# Trainer
trainer = Trainer(
    model=model, args=args,
    train_dataset=train_ds,
    tokenizer=tok,
    data_collator=collator,
    compute_metrics=compute_metrics
)

# Train
trainer.train()

# Quick train reference score
logits_tr = trainer.predict(train_ds).predictions
print("[Model] Train macro-F1:", f1_score(train_df["label"], logits_tr.argmax(-1), average="macro"))

# Predict on test
logits_te = trainer.predict(test_ds).predictions
proba_electra = torch.softmax(torch.tensor(logits_te), dim=1).numpy()
pred_electra  = proba_electra.argmax(1)
```
